```python
import requests
import pandas as pd

url = "https://api.census.gov/data/2022/acs/acs5"
params = {
    "get": "B19013_001E,NAME",
    "for": "zip code tabulation area:*",
    "in": "state:36"  # 36 = New York
}
r = requests.get(url, params=params)
income_df = pd.DataFrame(r.json()[1:], columns=r.json()[0])
income_df = income_df.rename(columns={"B19013_001E": "median_income", "zip code tabulation area": "zcta"})
income_df.to_csv("data/nyc_income_by_zcta.csv", index=False)
```
